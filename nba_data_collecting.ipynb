```python
def profile_dataframe(df, name):
    null_counts = df.isnull().sum()
    summary = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.values,
        "null_count": null_counts.values,
        "null_pct": (null_counts.values / len(df) * 100).round(2),
    })
    print(f"--- {name} ---")
    print(f"rows: {len(df)}, columns: {len(df.columns)}")
    print(summary.to_string(index=False))
    return summary
```

### `league_stats` — column reference

**Identity columns:**
- **PLAYER_ID** (`int64`) — unique player ID, matches `id` in `players_df`
- **PLAYER_NAME** (`object`) — player's full name
- **NICKNAME** (`object`) — first name / shorthand name used on broadcasts
- **TEAM_ID** (`int64`) — unique team ID, matches `id` in `teams_df`
- **TEAM_ABBREVIATION** (`object`) — 3-letter team code
- **AGE** (`float64`) — player's age as of this season

**Season totals (per player, 2025-26 regular season):**
- **GP** — games played, **W**/**L** — team wins/losses in games this player appeared in, **W_PCT** — win percentage
- **MIN** — total minutes played
- Shooting: **FGM/FGA/FG_PCT** (field goals made/attempted/%), **FG3M/FG3A/FG3_PCT** (3-pointers), **FTM/FTA/FT_PCT** (free throws)
- Rebounding: **OREB** (offensive), **DREB** (defensive), **REB** (total)
- **AST** — assists, **TOV** — turnovers, **STL** — steals, **BLK** — blocks made, **BLKA** — blocks *against* (shots blocked while shooting)
- **PF** — personal fouls committed, **PFD** — personal fouls drawn
- **PTS** — total points, **PLUS_MINUS** — team's point differential while this player was on court
- **NBA_FANTASY_PTS** / **WNBA_FANTASY_PTS** — fantasy scoring totals under each league's scoring formula
- **DD2** — double-doubles, **TD3** — triple-doubles

**Rank columns:** every stat above has a matching `<STAT>_RANK` column (e.g. `PTS_RANK`, `AST_RANK`) — the player's league-wide rank for that stat this season. That's 30 of the 67 columns, all following this same suffix pattern rather than needing individual descriptions.

**TEAM_COUNT** (`int64`) — number of different teams this player appeared for this season (>1 means they were traded).

### `draft_history` — column reference

- **PERSON_ID** (`int64`) — unique player ID, matches `id` in `players_df`
- **PLAYER_NAME** (`object`) — player's full name at time of draft
- **SEASON** (`object`) — draft year
- **ROUND_NUMBER** (`int64`) — which round the player was picked in (1 or 2 in the modern draft; historically drafts had more rounds)
- **ROUND_PICK** (`int64`) — pick number within that round
- **OVERALL_PICK** (`int64`) — pick number overall across the whole draft
- **DRAFT_TYPE** (`object`) — type of draft event (e.g. standard "Draft")
- **TEAM_ID** (`int64`) — drafting team's ID, matches `id` in `teams_df`
- **TEAM_CITY** / **TEAM_NAME** / **TEAM_ABBREVIATION** (`object`) — drafting team's city, name, and 3-letter code at the time
- **ORGANIZATION** (`object`) — the college, international team, or organization the player came from
- **ORGANIZATION_TYPE** (`object`) — category of that organization (e.g. "College", "International")
- **PLAYER_PROFILE_FLAG** (`int64`) — flag indicating whether this player has a full profile page in the NBA's system (older/obscure picks may not)

Rows: 8,374 (every draft pick recorded, spanning nearly 80 years). Columns: 14. Nulls: 0 across all columns — draft records are administrative bookkeeping, complete regardless of era, unlike in-game stat tracking which has known historical gaps.

### `league_leaders` — column reference

- **PLAYER_ID** (`int64`) — unique player ID, matches `id` in `players_df`
- **RANK** (`int64`) — player's overall rank for this leaderboard's default sort stat (points, by default)
- **PLAYER** (`object`) — player's full name
- **TEAM_ID** (`int64`) — team ID, matches `id` in `teams_df`
- **TEAM** (`object`) — 3-letter team abbreviation
- **GP** (`int64`) — games played
- **MIN** (`int64`) — total minutes played
- **FGM/FGA/FG_PCT**, **FG3M/FG3A/FG3_PCT**, **FTM/FTA/FT_PCT** — season shooting totals (made/attempted/percentage, for field goals, 3-pointers, free throws)
- **OREB/DREB/REB** (`int64`) — offensive, defensive, total rebounds
- **AST** (`int64`) — assists, **STL** — steals, **BLK** — blocks, **TOV** — turnovers, **PF** — personal fouls
- **PTS** (`int64`) — total points scored
- **EFF** (`int64`) — NBA efficiency rating, a single combined-value stat: `(PTS + REB + AST + STL + BLK) - ((FGA - FGM) + (FTA - FTM) + TOV)`
- **AST_TOV** (`float64`) — assist-to-turnover ratio (higher = more careful with the ball relative to playmaking)
- **STL_TOV** (`float64`) — steal-to-turnover ratio

Rows: 582 (every player with recorded stats this season). Columns: 28. Nulls: 0 across all columns — season-aggregated totals, same pattern as every other dashboard-style table in this notebook.
